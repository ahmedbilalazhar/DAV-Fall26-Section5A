##### Task 3 — Handle Missing Values

```python
df['workclass'].value_counts()
```

```python
df['occupation'].value_counts()
```

```python
df['workclass'] = df['workclass'].fillna('Unknown')
df['occupation'] = df['occupation'].fillna('Unknown')
df['native.country'] = df['native.country'].fillna(df['native.country'].mode()[0])
df.isnull().sum()
```

```python
df[['age', 'fnlwgt', 'education.num', 'capital.gain', 'capital.loss', 'hours.per.week']].isnull().sum()
```

**Justification:** `workclass` and `occupation` are imputed with an `'Unknown'` category instead of being dropped, since they make up a meaningful share of rows and dropping them would throw away usable data in every other column too. `native.country` is imputed with its mode (`United-States`) since missing values are a small share of a heavily one-sided column, so imputing barely changes the distribution. None of the numeric columns have missing values in this dataset once the `'?'` placeholders are fixed, so no numeric imputation is needed.

##### Task 10 — Data Profiling Summary Report

###### Dataset Overview
The raw dataset contains 32,561 rows and 15 columns; after removing exact duplicates the cleaned dataset has **32,537 rows** and **15 columns**. It is the UCI Adult Income (Census Income) dataset — each row represents an individual from the 1994 US Census database, with demographic and employment attributes (age, workclass, education, occupation, marital status, hours worked per week, etc.) used to predict whether that person's annual income is above or below $50K.

###### Data Quality Issues Found
- **Missing values:** Hidden as the string `'?'` rather than true `NaN`, so `.isnull().sum()` initially reported zero missing values. After replacement: `occupation` had 1,843 missing, `workclass` had 1,836 missing, and `native.country` had 583 missing.
- **Duplicates:** 24 exact duplicate rows were found and removed. Checking duplicates while ignoring the `income` column found 25 matches (one additional row pair that only differs on `income`); these were kept rather than dropped, since they can plausibly represent different individuals with identical recorded attributes but different real outcomes.
- **Inconsistent data:** Categorical text columns (e.g., `education`, `marital.status`, `native.country`, `occupation`) contained leading/trailing whitespace, which was stripped.

###### Key Observations
- `income` classes are imbalanced: ~75.9% `<=50K` vs ~24.1% `>50K`, which matters for any downstream classification modeling (accuracy alone would be misleading).
- `sex` distribution is imbalanced: ~66.9% Male vs ~33.1% Female.
- The most common occupation is `Prof-specialty` (4,136 records), closely followed by `Craft-repair` and `Exec-managerial`.
- Mean age (~38.6) is close to median age (37); mean hours-per-week (~40.4) is close to median (40) — both distributions are roughly symmetric with only mild right skew from high-hour/older outliers.
- `education` and `education.num` are fully consistent — every education label maps to exactly one `education.num` value, with zero inconsistencies found.

###### Cleaning Decisions Made
- Replaced `'?'` placeholder strings with proper `NaN`.
- Imputed `workclass` and `occupation` missing values with an `'Unknown'` category rather than dropping rows, to retain the rest of each record's information instead of losing ~11% of rows.
- Imputed `native.country` missing values with its mode (`United-States`), since it is a small share (~1.8%) of a heavily skewed column.
- Dropped the 24 exact duplicate rows; retained rows that matched only on every column except the `income` label.
- Stripped whitespace from all categorical text columns to eliminate false "inconsistencies" in category values.
